```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("train.csv")
df.head()
```

```python
df['Survived'].value_counts().plot(kind='bar')
plt.title("Survival Count")
plt.xlabel("0 = Died, 1 = Survived")
plt.ylabel("Number of People")
plt.show()
```

```python
sns.countplot(x='Pclass', hue='Sex', data=df)
plt.title("Passengers Distribution by Class and Gender")
plt.xlabel("Passenger Class (1 = Upper, 2 = Middle, 3 = Lower)")
plt.ylabel("Number of Passengers")
plt.show()

sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Survival by Passenger Class")
plt.xlabel("0 = Died, 1 = Survived")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
#Pandas

# What is the difference between the Pandas series and dataframe?
# Series is a one-dimensional data structure that contains a single column of data.
# DataFrame is a two-dimensional data structure that contains rows and multiple columns.

# Delete or Add column in Dataframe. (Also define concept of Axis)
# A new column can be added by assigning values, and a column can be deleted using drop().
# Axis defines direction: axis=0 represents rows and axis=1 represents columns.

df['New'] = 1
df.drop('New', axis=1, inplace=True)


# How do we change column names in Dataframe?
# Column names can be changed using the rename() function.

df.rename(columns={'Age': 'MyAge'}, inplace=True)


# Apply logical conditions to print values greater than certain threshold in any one column.
# Logical conditions are used to filter data based on a condition.

df[df['Age'] > 30]

# How do I access data in a dataframe? (Explore loc and iloc options)
# loc is label-based indexing and iloc is position-based indexing.

df.loc[0]
df.iloc[0]


# Create Excel file with multiple sheets and than load 3rd sheet in Excel file
# Multiple sheets can be created using ExcelWriter and a specific sheet can be loaded using read_excel().

with pd.ExcelWriter('output.xlsx') as writer:
    df.to_excel(writer, sheet_name='Sheet1')
    df.to_excel(writer, sheet_name='Sheet2')
    df.to_excel(writer, sheet_name='Sheet3')

df3 = pd.read_excel('output.xlsx', sheet_name='Sheet3')
df3.head()
```
