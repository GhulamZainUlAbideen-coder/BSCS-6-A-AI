```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Pakistan Largest Ecommerce Dataset.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.fillna(0)
```

```python
df = df.drop_duplicates()
```

```python
df = df.drop(['Unnamed: 0'], axis=1, errors='ignore')
```

```python
df['category_name_1'].value_counts()
```

```python
sns.countplot(x='payment_method', data=df)
plt.title("Payment Methods")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(x='status', data=df)
plt.title("Order Status")
plt.xlabel("Status")
plt.ylabel("Count")
plt.show()
```

```python
pd.crosstab(df['payment_method'], df['status'])
```

```python
df.to_csv("cleaned_dataset.csv", index=False)
```

```python
# Missing Values:
# Missing values were replaced with 0 for simplicity.

# Duplicate Values:
# Duplicate rows were removed.

# Outliers:
# No complex outlier detection method was used, simple cleaning applied.

# Categorical Data:
# Categorical data was kept as it is for simplicity.
```
